```python
import pandas as pd
import numpy as np
import os
df = pd.read_csv("../dataset/Indian_healthcare.csv")
print("Original Dataset Shape:", df.shape)
data = df.copy()


data["Treatment_Cost_INR"] = 3000


diagnosis_cost = {
    "Diabetes": 5000,
    "Hypertension": 4500,
    "Heart Disease": 15000,
    "Asthma": 7000,
    "Infection": 4000,
    "Fever": 2500,
    "Kidney Disease": 12000,
    "Liver Disease": 11000,
    "Cancer": 25000,
    "Other": 5000
}

if "Primary_Diagnosis" in data.columns:
    data["Treatment_Cost_INR"] += (
        data["Primary_Diagnosis"]
        .map(diagnosis_cost)
        .fillna(5000)
    )



treatment_cost = {
    "Medication": 2000,
    "Surgery": 20000,
    "Therapy": 8000,
    "Consultation": 1500,
    "Emergency": 12000,
    "Other": 5000
}

if "Treatment_Type" in data.columns:
    data["Treatment_Cost_INR"] += (
        data["Treatment_Type"]
        .map(treatment_cost)
        .fillna(5000)
    )



imaging_cost = {
    "MRI": 5000,
    "CT Scan": 4000,
    "X-Ray": 1500,
    "Ultrasound": 2000,
    "None": 0
}

if "Imaging_Type" in data.columns:
    data["Treatment_Cost_INR"] += (
        data["Imaging_Type"]
        .map(imaging_cost)
        .fillna(1000)
    )


hospital_cost = {
    "Private": 6000,
    "Government": 2000,
    "Clinic": 1500
}

if "Hospital_Type" in data.columns:
    data["Treatment_Cost_INR"] += (
        data["Hospital_Type"]
        .map(hospital_cost)
        .fillna(3000)
    )


if "Age" in data.columns:
    data["Treatment_Cost_INR"] += np.where(
        data["Age"] >= 60,
        4000,
        np.where(data["Age"] >= 40, 2000, 0)
    )


if "BMI" in data.columns:
    data["Treatment_Cost_INR"] += np.where(
        data["BMI"] >= 30,
        2500,
        np.where(data["BMI"] >= 25, 1000, 0)
    )


if "Blood_Glucose_mg_dL" in data.columns:
    data["Treatment_Cost_INR"] += np.where(
        data["Blood_Glucose_mg_dL"] >= 200,
        3000,
        np.where(data["Blood_Glucose_mg_dL"] >= 126, 1500, 0)
    )


if "Insurance_Covered" in data.columns:
    data["Treatment_Cost_INR"] += np.where(
        data["Insurance_Covered"].astype(str).str.lower() == "true",
        500,
        1500
    )


np.random.seed(42)

variation = np.random.normal(
    loc=0,
    scale=1500,
    size=len(data)
)

data["Treatment_Cost_INR"] += variation


data["Treatment_Cost_INR"] = data["Treatment_Cost_INR"].clip(
    lower=1000
)

data["Treatment_Cost_INR"] = data["Treatment_Cost_INR"].round(2)


def cost_category(cost):

    if cost < 10000:
        return "Low"

    elif cost < 25000:
        return "Medium"

    else:
        return "High"


data["Cost_Category"] = data["Treatment_Cost_INR"].apply(
    cost_category
)



print("\n========== SAMPLE DATA ==========")

columns_to_show = [
    "Age",
    "Primary_Diagnosis",
    "Treatment_Type",
    "Hospital_Type",
    "Treatment_Cost_INR",
    "Cost_Category"
]

print(data[columns_to_show].head(10))


print("\n========== COST STATISTICS ==========")

print(data["Treatment_Cost_INR"].describe())


print("\n========== COST CATEGORY ==========")

print(data["Cost_Category"].value_counts())


os.makedirs("../outputs", exist_ok=True)

data.to_csv(
    "../outputs/healthcare_cost_dataset.csv",
    index=False
)

print("\nDataset saved successfully!")
print("../outputs/healthcare_cost_dataset.csv")

print("\nFinal Dataset Shape:", data.shape)
```
