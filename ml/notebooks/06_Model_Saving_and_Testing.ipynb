```python
import pandas as pd
import joblib
import os


model_path = "../outputs/best_medical_cost_model.pkl"

model = joblib.load(model_path)

print("Model loaded successfully!")


new_patient = pd.DataFrame([{

    "Age": 45,

    "Gender": "Female",

    "Region": "South",

    "Socioeconomic_Status": "Middle",

    "Primary_Diagnosis": "Diabetes",

    "Blood_Glucose_mg_dL": 160,

    "HbA1c_%": 7.2,

    "Total_Cholesterol_mg_dL": 210,

    "Treatment_Type": "Medication",

    "Treatment_Outcome": "Improved",

    "Imaging_Type": "None",

    "Hospital_Type": "Private",

    "Insurance_Covered": True,

    "BMI": 28.5

}])


print("\n========== NEW PATIENT ==========")

print(new_patient)


predicted_cost = model.predict(new_patient)[0]


if predicted_cost < 10000:

    category = "Low"

elif predicted_cost < 25000:

    category = "Medium"

else:

    category = "High"


print("\n========== PREDICTION RESULT ==========")

print(
    "Estimated Medical Cost: ₹",
    round(predicted_cost, 2)
)

print(
    "Cost Category:",
    category
)


test_patients = pd.DataFrame([

    {
        "Age": 25,
        "Gender": "Female",
        "Region": "South",
        "Socioeconomic_Status": "Middle",
        "Primary_Diagnosis": "Fever",
        "Blood_Glucose_mg_dL": 95,
        "HbA1c_%": 5.2,
        "Total_Cholesterol_mg_dL": 170,
        "Treatment_Type": "Medication",
        "Treatment_Outcome": "Improved",
        "Imaging_Type": "None",
        "Hospital_Type": "Government",
        "Insurance_Covered": True,
        "BMI": 22.5
    },

    {
        "Age": 65,
        "Gender": "Male",
        "Region": "South",
        "Socioeconomic_Status": "High",
        "Primary_Diagnosis": "Heart Disease",
        "Blood_Glucose_mg_dL": 180,
        "HbA1c_%": 8.0,
        "Total_Cholesterol_mg_dL": 240,
        "Treatment_Type": "Surgery",
        "Treatment_Outcome": "Improved",
        "Imaging_Type": "CT Scan",
        "Hospital_Type": "Private",
        "Insurance_Covered": False,
        "BMI": 31.0
    }

])


predictions = model.predict(test_patients)


print("\n========== MULTIPLE PATIENT TEST ==========")

for i, cost in enumerate(predictions):

    if cost < 10000:
        category = "Low"

    elif cost < 25000:
        category = "Medium"

    else:
        category = "High"

    print(
        f"Patient {i + 1}: "
        f"₹{cost:.2f} - {category}"
    )



print("\n======================================")
print("MODEL TESTING COMPLETED SUCCESSFULLY")
print("======================================")

print("\nModel file:")
print(model_path)
```
