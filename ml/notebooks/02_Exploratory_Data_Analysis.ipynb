```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("../dataset/Indian_healthcare.csv")


print("Dataset Shape:", df.shape)


plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()


plt.figure(figsize=(8,5))
plt.hist(df["BMI"], bins=20)
plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Number of Patients")
plt.show()

# Blood glucose distribution
plt.figure(figsize=(8,5))
plt.hist(df["Blood_Glucose_mg_dL"], bins=20)
plt.title("Blood Glucose Distribution")
plt.xlabel("Blood Glucose (mg/dL)")
plt.ylabel("Number of Patients")
plt.show()


plt.figure(figsize=(8,5))
plt.hist(df["HbA1c_%"], bins=20)
plt.title("HbA1c Distribution")
plt.xlabel("HbA1c (%)")
plt.ylabel("Number of Patients")
plt.show()


plt.figure(figsize=(10,5))
df["Primary_Diagnosis"].value_counts().plot(kind="bar")
plt.title("Primary Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.show()


plt.figure(figsize=(8,5))
df["Treatment_Type"].value_counts().plot(kind="bar")
plt.title("Treatment Type")
plt.xlabel("Treatment")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.show()


plt.figure(figsize=(8,5))
df["Hospital_Type"].value_counts().plot(kind="bar")
plt.title("Hospital Type")
plt.xlabel("Hospital Type")
plt.ylabel("Number of Patients")
plt.show()

# Insurance coverage
plt.figure(figsize=(8,5))
df["Insurance_Covered"].value_counts().plot(kind="bar")
plt.title("Insurance Coverage")
plt.xlabel("Insurance Covered")
plt.ylabel("Number of Patients")
plt.show()


numerical_columns = [
    "Age",
    "Blood_Glucose_mg_dL",
    "HbA1c_%",
    "Total_Cholesterol_mg_dL",
    "BMI"
]

print("\nCorrelation Matrix:")
print(df[numerical_columns].corr())


print("\nEDA Completed Successfully!")
```
