```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import os


df = pd.read_csv("../indian_healthcare_patient_records.csv")

print("Original Shape:", df.shape)


if "Patient_ID" in df.columns:
    df = df.drop("Patient_ID", axis=1)


if "Visit_Date" in df.columns:
    df["Visit_Date"] = pd.to_datetime(df["Visit_Date"])
    df["Visit_Year"] = df["Visit_Date"].dt.year
    df["Visit_Month"] = df["Visit_Date"].dt.month
    df = df.drop("Visit_Date", axis=1)


text_columns = [
    "Symptoms",
    "Imaging_Findings",
    "Occupation"
]

for column in text_columns:
    if column in df.columns:
        df = df.drop(column, axis=1)


categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns

encoder = LabelEncoder()

for column in categorical_columns:
    df[column] = encoder.fit_transform(
        df[column].astype(str)
    )


print("\nProcessed Dataset:")
print(df.head())

print("\nProcessed Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


os.makedirs("../outputs", exist_ok=True)

df.to_csv(
    "../outputs/processed_healthcare_data.csv",
    index=False
)

print("\nPreprocessing Completed Successfully!")
print("Saved as: ../outputs/processed_healthcare_data.csv")
```
